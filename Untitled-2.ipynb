```python
import pandas as pd

# Read the sales transactions file
df_sales = pd.read_csv("Sales_Transactions.csv")

# 1. Calculate total gross sales, net sales, gross profit, total units, and average margin
total_gross_sales = df_sales["Gross_Sales_EGP"].sum()
total_net_sales = df_sales["Net_Sales_EGP"].sum()
total_profit = df_sales["Gross_Profit_EGP"].sum()
total_units = df_sales["Quantity"].sum()
avg_margin = df_sales["Gross_Margin_Pct"].mean() * 100

print("=== EXECUTIVE SALES & PROFITABILITY REPORT ===")
print(f"Total Gross Sales: {total_gross_sales:,.2f} EGP")
print(f"Total Net Sales: {total_net_sales:,.2f} EGP")
print(f"Total Gross Profit: {total_profit:,.2f} EGP")
print(f"Total Units Sold: {total_units:,} units")
print(f"Average Gross Margin: {avg_margin:.2f}%")

# 2. Performance breakdown by Sales Channel
channel_perf = (
    df_sales.groupby("Channel")
    .agg(
        Net_Sales=("Net_Sales_EGP", "sum"),
        Gross_Profit=("Gross_Profit_EGP", "sum"),
        Total_Units=("Quantity", "sum"),
    )
    .reset_index()
)

print("\n=== Performance by Sales Channel ===")
print(channel_perf)
```

```python
import pandas as pd

# قراءة بيانات العملاء والمبيعات
df_customers = pd.read_csv("Customers.csv")
df_sales = pd.read_csv("Sales_Transactions.csv")

print("=== CUSTOMER ANALYSIS REPORT ===")
print(f"Total Unique Customers in Database: {df_customers['Customer_ID'].nunique():,}")

# تحليل العملاء بناءً على المبيعات (متوسط قيمة الطلب لكل عميل)
customer_spend = df_sales.groupby("Customer_ID").agg(
    Total_Spent=("Net_Sales_EGP", "sum"),
    Total_Orders=("Invoice_ID", "nunique"),
    Average_Order_Value=("Net_Sales_EGP", "mean")
).reset_index()

print("\nTop 5 Customers by Spending:")
print(customer_spend.sort_values(by="Total_Spent", ascending=False).head())
```

```python
import pandas as pd

df_sales = pd.read_csv("Sales_Transactions.csv")

print("=== BRANCH & CHANNEL PERFORMANCE ===")

# 1. Performance by Location/Branch
branch_perf = df_sales.groupby("Location_Name").agg(
    Total_Net_Sales=("Net_Sales_EGP", "sum"),
    Total_Returns=("Returned", lambda x: (x == "Yes").sum()),
    Average_Margin=("Gross_Margin_Pct", "mean")
).reset_index()

print("\nPerformance by Branch / Location:")
print(branch_perf.sort_values(by="Total_Net_Sales", ascending=False))
```

```python
import pandas as pd

# Read factory production data
df_prod = pd.read_csv("Production_Orders.csv")

print("=== FACTORY PERFORMANCE SUMMARY ===")

# 1. Average efficiency by production line
line_efficiency = (
    df_prod.groupby("Line_ID")["Efficiency_Pct"].mean().reset_index()
)
print("\nAverage Efficiency by Line (%):")
print(line_efficiency)

# 2. Total rejected (damaged) quantities by line
line_rejections = (
    df_prod.groupby("Line_ID")["Rejected_Qty"].sum().reset_index()
)
print("\nTotal Rejected Quantities by Line:")
print(line_rejections)

# 3. Total overtime hours by line
line_overtime = (
    df_prod.groupby("Line_ID")["Overtime_Hours"].sum().reset_index()
)
print("\nTotal Overtime Hours by Line:")
print(line_overtime)
```

```python
import pandas as pd

# قراءة بيانات المشتريات
df_po = pd.read_csv("Purchase_Orders.csv")

print("=== SUPPLIER PERFORMANCE SUMMARY ===")

# 1. إجمالي المشتريات والكميات المرفوضة لكل مورد
supplier_rejections = (
    df_po.groupby("Supplier_ID")
    .agg(Total_Orders=("PO_ID", "count"), Total_Rejected=("Rejected_Qty", "sum"))
    .reset_index()
)
print("\nRejected Quantities by Supplier:")
print(supplier_rejections)
```

```python
import pandas as pd

# قراءة بيانات خدمة العملاء
df_cs = pd.read_csv("Customer_Service.csv")

print("=== CUSTOMER SERVICE SUMMARY ===")

# 1. أكثر أنواع المشاكل تكراراً
issue_counts = df_cs["Issue_Type"].value_counts().reset_index()
issue_counts.columns = ["Issue_Type", "Count"]
print("\nMost Common Issues:")
print(issue_counts)

# 2. متوسط تقييم العملاء (CSAT) حسب القناة
csat_by_channel = df_cs.groupby("Channel")["CSAT_1_to_5"].mean().reset_index()
print("\nAverage CSAT by Channel:")
print(csat_by_channel)
```
